```python
full_missing = pd.read_csv(RAW_DATA_PATH, low_memory=False).isna().mean().sort_values(ascending=False)

full_missing.head(40)
```

```python
missing_counts = None
total_rows = 0

for chunk in pd.read_csv(RAW_DATA_PATH, chunksize=100_000, low_memory=False):
    if missing_counts is None:
        missing_counts = chunk.isna().sum()
    else:
        missing_counts += chunk.isna().sum()
    total_rows += len(chunk)

missing_fraction = (missing_counts / total_rows).sort_values(ascending=False)

missing_fraction.head(40)
```

```python
missing_fraction.to_csv("../results/tables/missing_fraction_by_column.csv")
```
